```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        
    def __str__(self):
        return f"Person(name={self.name}, age={self.age})"
    
    def __repr__(self):
        return f"Person(name={self.name}, age={self.age})"
    
    def __eq__(self, other):
        return self.name == other.name and self.age == other.age
    
    def __hash__(self):
        return hash((self.name, self.age))
    
    def __lt__(self, other):
        return self.age < other.age
    
    def __le__(self, other):
        return self.age <= other.age
    
    def __gt__(self, other):
        return self.age > other.age 
    
    def __ge__(self, other):
        return self.age >= other.age        
    
    def __add__(self, other):
        return Person(f"{self.name} {other.name}", self.age + other.age)
    
    def __mul__(self, other):
        return Person(f"{self.name} {other.name}", self.age * other.age)
    
    def __pow__(self, other):
        return Person(f"{self.name} {other.name}", self.age ** other.age)
    
    def __truediv__(self, other):
        return Person(f"{self.name} {other.name}", self.age / other.age)    
    
    def __floordiv__(self, other):    
        return Person(f"{self.name} {other.name}", self.age // other.age)
    
    def __mod__(self, other):   
        return Person(f"{self.name} {other.name}", self.age % other.age)    
    
    def __and__(self, other):   
        return Person(f"{self.name} {other.name}", self.age & other.age)    
        
    def __or__(self, other):   
        return Person(f"{self.name} {other.name}", self.age | other.age)    
        
    def __xor__(self, other):   
        return Person(f"{self.name} {other.name}", self.age ^ other.age)    
        
    def __lshift__(self, other):   
        return Person(f"{self.name} {other.name}", self.age << other.age)    
        
    def __rshift__(self, other):   
        return Person(f"{self.name} {other.name}", self.age >> other.age)    
        
    def __neg__(self):   
        return Person(f"{self.name} {other.name}", -self.age)    
        
    def __pos__(self):   
        return Person(f"{self.name} {other.name}", +self.age)    
        
    def __abs__(self):   
        return Person(f"{self.name} {other.name}", abs(self.age))    
        
    def __invert__(self):   
        return Person(f"{self.name} {other.name}", ~self.age)    
        
    def __complex__(self):   
        return Person(f"{self.name} {other.name}", complex(self.age))    
        
    def __int__(self):   
        return Person(f"{self.name} {other.name}", int(self.age))    
        
    def __float__(self):   
        return Person(f"{self.name} {other.name}", float(self.age))    
        
    def __complex__(self):   
        return Person(f"{self.name} {other.name}", complex(self.age))   
        
    def __index__(self):   
        return Person(f"{self.name} {other.name}", self.age)
        
    def __bool__(self):   
        return Person(f"{self.name} {other.name}", bool(self.age))
        
    def __bytes__(self):   
        return Person(f"{self.name} {other.name}", bytes(self.age))
        
    def __format__(self):   
        return Person(f"{self.name} {other.name}", format(self.age))
        
    def __round__(self):   
        return Person(f"{self.name} {other.name}", round(self.age))
```
